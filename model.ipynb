```python
X=df.drop("median_house_value",axis=1)
```

```python
Y=df["median_house_value"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=0.2,
    random_state=78
)
```

```python
print(X_train)
```

```python
from sklearn.impute import SimpleImputer
```

```python
imputer=SimpleImputer(strategy="median")
```

```python
X_train[["total_bedrooms"]]=imputer.fit_transform(
    X_train[["total_bedrooms"]]
)
X_test[["total_bedrooms"]]=imputer.transform(
    X_test[["total_bedrooms"]]
)
```

```python
print(X_train["total_bedrooms"].isnull().sum())
```

```python
print(X_test["total_bedrooms"].isnull().sum())
```

```python
print(imputer.statistics_)
```

```python
print(X_train.info())
```

```python
print(X_train["ocean_proximity"].unique())
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
ocean_train=encoder.fit_transform(
    X_train[["ocean_proximity"]]
)
```

```python
print(ocean_train[:5])
```

```python
ocean_test=encoder.transform(
    X_test[["ocean_proximity"]]
)
```

```python
ocean_columns=encoder.get_feature_names_out(
    ["ocean_proximity"]
)
print(ocean_columns)
```

```python
print(X_train.shape)
print(X_test.shape)

print(ocean_train.shape)
print(ocean_test.shape)
```

```python
ocean_train_df = pd.DataFrame(
    ocean_train,
    columns=ocean_columns,
    index=X_train.index
)

ocean_test_df = pd.DataFrame(
    ocean_test,
    columns=ocean_columns,
    index=X_test.index
)
```

```python
X_train=X_train.drop("ocean_proximity",axis=1)
X_test=X_test.drop("ocean_proximity",axis=1)
```

```python
X_train=pd.concat(
    [X_train,ocean_train_df],axis=1
)
X_test=pd.concat(
    [X_test,ocean_test_df],axis=1
)
```

```python
print(X_train.head())
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler=StandardScaler()
```

```python
numeric_columns=[
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income"
]
```

```python
X_train[numeric_columns]=scaler.fit_transform(
    X_train[numeric_columns]
)
```

```python
X_test[numeric_columns]=scaler.transform(
    X_test[numeric_columns]
)
```

```python
print(X_train[numeric_columns].mean())
```

```python
print(X_train[numeric_columns].std())
```

```python
from sklearn.linear_model import LinearRegression
```

```python
model=LinearRegression()
```

```python
model.fit(X_train,Y_train)
```

```python
print(model.intercept_)
```

```python
print(model.coef_)
```

```python
predictions=model.predict(X_test)
```

```python
print(predictions[:10])
```

```python
print(Y_test.iloc[:10].values)
```

```python
comparision=pd.DataFrame({
    "Acutal":Y_test.values,
    "Predicted":predictions
})
print(comparision.head(10))
```

```python
comparision["Error"]=(
    comparision["Acutal"]-comparision["Predicted"]
)
print(comparision.head(10))
```

```python
from sklearn.metrics import mean_absolute_error
```

```python
mae=mean_absolute_error(
    Y_test,
    predictions
)
```

```python
print("MAE:",mae)
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mse=mean_squared_error(
    Y_test,
    predictions
)
print("MSE:",mse)
```

```python
import numpy as np
```

```python
rmse=np.sqrt(mse)
```

```python
print("RMSE:",rmse)
```

```python
from sklearn.metrics import r2_score
```

```python
r2=r2_score(
    Y_test,
    predictions
)
print("R^2:",r2)
```

```python
import numpy as np

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


mae = mean_absolute_error(
    Y_test,
    predictions
)

mse = mean_squared_error(
    Y_test,
    predictions
)

rmse = np.sqrt(mse)

r2 = r2_score(
    Y_test,
    predictions
)


print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
print("R²:", r2)
```
