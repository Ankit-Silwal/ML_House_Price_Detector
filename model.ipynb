```python
from sklearn.preprocessing import OneHotEncoder
```

```python
encoder=OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
ocean_train=encoder.fit_transform(
    X_train[["ocean_proximity"]]
)
```

```python
print(ocean_train[:5])
```

```python
ocean_test=encoder.transform(
    X_test[["ocean_proximity"]]
)
```

```python
ocean_columns=encoder.get_feature_names_out(
    ["ocean_proximity"]
)
print(ocean_columns)
```

```python
print(X_train.shape)
print(X_test.shape)

print(ocean_train.shape)
print(ocean_test.shape)
```

```python
ocean_train_df = pd.DataFrame(
    ocean_train,
    columns=ocean_columns,
    index=X_train.index
)

ocean_test_df = pd.DataFrame(
    ocean_test,
    columns=ocean_columns,
    index=X_test.index
)
```

```python
X_train=X_train.drop("ocean_proximity",axis=1)
X_test=X_test.drop("ocean_proximity",axis=1)
```

```python
X_train=pd.concat(
    [X_train,ocean_train_df],axis=1
)
X_test=pd.concat(
    [X_test,ocean_test_df],axis=1
)
```
